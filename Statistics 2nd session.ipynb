```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom


n = 20
p = 0.35
x = np.arange(0, n + 1)

pmf = binom.pmf(x, n, p)

cdf = binom.cdf(x, n, p)

mean, var = binom.stats(n, p, moments='mv')  # mean = 7.0, variance = 4.55

samples = binom.rvs(n, p, size=1000)

plt.figure(figsize=(8, 4))
plt.bar(x, pmf, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Binomial Distribution PMF (n=20, p=0.35)')
plt.xlabel('Number of Successes (x)')
plt.ylabel('Probability P(X = x)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
